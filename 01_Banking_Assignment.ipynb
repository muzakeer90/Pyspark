```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.types import *

spark = SparkSession.builder.appName("BankingAssignment").getOrCreate()

data = [
    (1, 101, "Savings", "Deposit", 5000.0, "Nellore", "2026-09-01"),
    (2, 102, "Current", "Withdrawal", 2000.0, "Chennai", "2026-09-02"),
    (3, 103, "Savings", "Deposit", 7500.0, "Bangalore", "2026-09-03"),
    (4, 104, "Savings", "Transfer", 3000.0, "Hyderabad", "2026-09-04"),
    (5, 105, "Current", "Deposit", 10000.0, "Mumbai", "2026-09-05"),
    (6, 101, "Savings", "Withdrawal", 1500.0, "Nellore", "2026-09-06"),
    (7, 106, "Current", "Deposit", 4500.0, "Delhi", "2026-09-07"),
    (8, 107, "Savings", "Transfer", 2500.0, "Pune", "2026-09-08"),
    (9, 108, "Savings", "Deposit", 6000.0, "Chennai", "2026-09-09"),
    (10, 109, "Current", "Withdrawal", 3500.0, "Hyderabad", "2026-09-10"),
    (11, 110, "Savings", "Deposit", 8000.0, "Bangalore", "2026-09-11"),
    (12, 102, "Current", "Transfer", 4000.0, "Chennai", "2026-09-12"),
    (13, 111, "Savings", "Withdrawal", 1200.0, "Mumbai", "2026-09-13"),
    (14, 112, "Current", "Deposit", 9000.0, "Delhi", "2026-09-14"),
    (15, 113, "Savings", "Transfer", 5500.0, "Pune", "2026-09-15")
]

schema = StructType([
    StructField("transaction_id", IntegerType(), False),
    StructField("customer_id", IntegerType(), False),
    StructField("account_type", StringType(), False),
    StructField("transaction_type", StringType(), False),
    StructField("amount", DoubleType(), False),
    StructField("city", StringType(), False),
    StructField("transaction_date", StringType(), False)
])

df = spark.createDataFrame(data, schema)
```

```python
df.show(truncate=False)
```


#2. Print schema, columns and data types.

# 7. Use `collect` on a small selection of customer IDs.


```python
customer_ids = df.select("customer_id").limit(5).collect()
customer_ids
```

# 8. Use `countByValue` to count transaction types.


```python
transaction_type_counts = df.select("transaction_type").rdd.map(lambda r: r[0]).countByValue()
transaction_type_counts
```

# 9. Use `reduce` to calculate total transaction amount.


```python
total_amount = df.select("amount").rdd.map(lambda r: r[0]).reduce(lambda a, b: a + b)
total_amount
```

# 10. Check whether the DataFrame is empty.


```python
df.isEmpty()
```

# 11. Iterate through transaction ID and amount using `toLocalIterator`.


```python
for row in df.select("transaction_id", "amount").toLocalIterator():
    print(row)
```

# 12. Write the banking DataFrame to Parquet.

```python
df.write.mode("overwrite").parquet("banking_transactions")
```
