#Assignment 1 - File Reads

```python
#1. Read customers.csv
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
spark=SparkSession.builder.appName("Assignment1").getOrCreate()
df_customers=spark.read.csv("customers.csv")
df_customers.show()
```

```python
# 2. Read the same file with:
# o header
# o inferSchema

df_customers=spark.read.csv("customers.csv",header=True,inferSchema=True)
df_customers.show()
```

```python
df_orders=spark.read.csv("orders.csv",header=True,inferSchema=True)
df_orders.show()
df_orders.printSchema()
```

```python
#3. Read orders.csv using an explicitly defined schema.

from pyspark.sql.types import StructType, StructField, IntegerType, LongType, StringType, DateType, DoubleType

orders_schema = StructType([
    StructField("order_id", LongType(), True),
    StructField("customer_id", IntegerType(), True),
    StructField("order_date", DateType(), True),
    StructField("order_status", StringType(), True),
    StructField("payment_method", StringType(), True),
    StructField("shipping_city", StringType(), True),
    StructField("shipping_state", StringType(), True),
    StructField("total_amount", DoubleType(), True),
    StructField("discount_amount", DoubleType(), True),
    StructField("shipping_cost", DoubleType(), True)
])

df_orders = spark.read.schema(orders_schema).csv("orders.csv", header=True)
df_orders.show()
df_orders.printSchema()
```

```python
#4. Convert customers.csv to Parquet.
df_customers.write.parquet("customers.parquet")
```

```python
#5. Read the Parquet file.
df_customers_parquet = spark.read.parquet("customers.parquet")
df_customers_parquet.show()
```

#Assignment 3 — withColumn()

```python
df_orders.show()
```

```python
#1. Create net_amount:
from pyspark.sql.functions import col

df_orders = df_orders.withColumn("net_amount", col("total_amount") - col("discount_amount"))
df_orders.show(5)
```

```python
#2. Create final_amount:
df_orders = df_orders.withColumn("final_amount", col("net_amount") + col("shipping_cost"))
df_orders.show(5)
```

```python
#3. Create discount_percentage.
df_orders = df_orders.withColumn("discount_percentage", (col("discount_amount") / col("total_amount")) * 100)
df_orders.show(5)
```

```python
#4. Create order_year.
df_orders = df_orders.withColumn("order_year", year(col("order_date")))
df_orders.show(5)
```

```python
#5. Create order_month.
df_orders = df_orders.withColumn("order_month", month(col("order_date")))
df_orders.show(5)
```

```python
#6. Create order_quarter.
df_orders = df_orders.withColumn("order_quarter", quarter(col("order_date")))
df_orders.show(5)
```

```python
#7. Create order_week.
df_orders = df_orders.withColumn("order_week", weekofyear(col("order_date")))
df_orders.show(5)
```

```python
#8. Create is_high_value_order.
df_orders = df_orders.withColumn("is_high_value_order", col("total_amount") > 5000)
df_orders.show(5)
```

```python
#9. Create order_category:
# < 500      → Low
# 500-2000    → Medium
# > 2000      → High

from pyspark.sql.functions import when

df_orders = df_orders.withColumn(
    "order_category",
    when(col("total_amount") < 500, "Low")
    .when((col("total_amount") >= 500) & (col("total_amount") <= 2000), "Medium")
    .otherwise("High")
)
```

```python
#10. Create order_age_days based on today's date.
from pyspark.sql.functions import current_date

df_orders = df_orders.withColumn("order_age_days", datediff(current_date(), col("order_date")))
df_orders.show(5)
```

#Assignment 7 — Window Functions

```python
#1. Find the customer's first order.
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number,rank,dense_rank,sum,lag,lead,col
customer_window=Window.partitionBy("customer_id").orderBy("order_date")
df_orders.withColumn("row_number",row_number().over(customer_window)).filter(col("row_number")==1).show()
#
```

```python
#2. Find the customer's latest order.
df_orders.withColumn("row_number",row_number().over(customer_window.orderBy(col("order_date").desc()))).filter(col("row_number")==1).show()
```

```python
#3. Rank customers based on total spending.
df_orders.groupBy("customer_id").agg(sum("total_amount").alias("total_spending")).withColumn("rank",rank().over(Window.orderBy(col("total_spending").desc()))).show()
```

```python
#4. Rank orders within each customer.
df_orders.withColumn("rank",rank().over(customer_window)).show()
```

```python
#5. Find the top 3 orders for every customer.
df_orders.withColumn("order_rank",row_number().over(customer_window)).filter(col("order_rank")<=3).show()
```

```python
#6. Calculate running revenue for each customer.
df_orders.withColumn("running_revenue",sum("total_amount").over(customer_window)).show(7)
```

```python
#7. Calculate previous order amount using lag().
df_orders.withColumn("prev_amount", lag("total_amount").over(customer_window)).show(7)
```

```python
#8. Calculate next order amount using lead().
df_orders.withColumn("next_amount", lead("total_amount").over(customer_window)).show(8)
```

```python
#9. Calculate difference between current and previous order.
df_orders.withColumn('diff_prev_order', col('total_amount') - lag('total_amount').over(customer_window)).show(9)
#
```

```python
#10. Calculate cumulative order count for each customer.
df_orders.withColumn("cumulative_count", sum("total_amount").over(customer_window)).show(5)
```
