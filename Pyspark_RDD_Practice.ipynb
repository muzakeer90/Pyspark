Q1
Dataset
[5, 10, 15, 20, 25, 30]
Problem:
Create an RDD from the dataset and multiply every number by 2.
Expected Output
[10, 20, 30, 40, 50, 60]

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder .appName("RDD Practice") .getOrCreate()

sc = spark.sparkContext
```

```python
data = [5, 10, 15, 20, 25, 30]

rdd = sc.parallelize(data)

result = rdd.map(lambda x: x * 2)

print(result.collect())
```

Q50
Problem:
Perform the following complete analysis:
Calculate revenue for every order.
Keep only orders with revenue greater than 50000.
Calculate total revenue for each product.
Find the product with the highest revenue.
Save the final result into a single output partition.
Expected Output
Orders retained:
O1 -> Laptop -> 100000
O2 -> Mobile -> 100000
O6 -> Laptop -> 200000
Product Revenue:
Laptop -> 300000
Mobile -> 100000
Highest Revenue Product:
Laptop -> 300000
Output:
1 part file

```python
# Step 1: Calculate revenue
revenue=rdd.map(lambda x:(x[0],x[1], x[2]*x[3]))
# print(result.collect())
# Step 2: Keep revenue > 50000
filtered =revenue.filter(lambda x:x[2]>50000)
print("Orders retained:")
print(filtered.collect())
# Step 3: Product revenue
product_revenue=filtered.map(lambda x:(x[1],x[2])).reduceByKey(lambda x,y : x+y)
print("Product Revenue:")
print(product_revenue.collect())
# Step 4: Highest revenue product
highest=product_revenue.max(key=lambda x : x[1])
print("Highest Revenue Product:")
print(highest)
product_revenue.coalesce(1).saveAsTextFile(
    "output/final_result"
)
```
